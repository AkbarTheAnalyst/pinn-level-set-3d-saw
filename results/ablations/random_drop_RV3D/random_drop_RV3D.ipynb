# random_drop_RV3D

Drop a random 5% of points instead of the top 5%. Benchmark RV3D, 1 run(s), about 50 min on a T4 GPU.

Files needed in `MyDrive/SAW_ablation/`: `rv3d_core.py`, `saw_ablation.py`, `RV3D_reference_n201_T2.npz` (optional; computed and saved there if missing).

Results are saved to `MyDrive/SAW_ablation/random_drop_RV3D/` after every run; re-running the notebook skips finished runs.

```python
import os, sys, csv, time
import numpy as np, torch
from google.colab import drive
drive.mount('/content/drive')

PROJECT_DIR = '/content/drive/MyDrive/SAW_ablation'
sys.path.insert(0, PROJECT_DIR)
import rv3d_core as core
import saw_ablation as sa

EXP, BENCH = 'random_drop_RV3D', 'RV3D'
OUT = f'{PROJECT_DIR}/{EXP}'
os.makedirs(OUT, exist_ok=True)
device = 'cuda'
print(torch.cuda.get_device_name(0))
```

```python
RUNS = [{'mode': 'random', 'seed': 42, 'q': 0.95}]
CSV_PATH = f'{OUT}/{EXP}.csv'
grid = core.Grid(n=sa.GRID_N)
PHI_E = None
times, _ = core.snapshot_times(sa.BENCH[BENCH]['T'], sa.N_SNAP)
PHI_E = core.reference_fields(grid, times, sa.BENCH[BENCH]['T'], device, nsteps=sa.RK4_STEPS,
                              cache_path=f'{PROJECT_DIR}/RV3D_reference_n201_T2.npz')
```

```python
def key(r):
    return (r['mode'], str(r['seed']), str(r.get('q', '')), str(r.get('w_eik', '')))

done = set()
if os.path.exists(CSV_PATH):
    with open(CSV_PATH) as f:
        done = {(r['mode'], r['seed'], r['q'], r['w_eik']) for r in csv.DictReader(f)}
todo = [r for r in RUNS if key(r) not in done]
print(f'{len(RUNS) - len(todo)} done, {len(todo)} to go')

for run in todo:
    run = dict(run)
    print(run['mode'], 'seed', run['seed'], flush=True)
    row, tag = sa.run_one(core, BENCH, run, device, OUT, grid, PHI_E)
    new = not os.path.exists(CSV_PATH)
    with open(CSV_PATH, 'a', newline='') as f:
        w = csv.DictWriter(f, fieldnames=list(row.keys()))
        if new:
            w.writeheader()
        w.writerow(row)
    print(f"  rel L2 {row['avg_rel_L2']:.4f}%  MAPE {row['avg_vol_MAPE']:.4f}%  "
          f"final w {row['w_eik_final']}  {row['minutes']:.1f} min", flush=True)
```

```python
for r in csv.DictReader(open(CSV_PATH)):
    extra = f"  F_fill {float(r['avg_slot_filled']):.2f}%" if 'avg_slot_filled' in r else ''
    print(f"{r['mode']:7s} seed {r['seed']}  rel L2 {float(r['avg_rel_L2']):.4f}%  "
          f"MAPE {float(r['avg_vol_MAPE']):.4f}%{extra}  final w {r['w_eik_final']}")
```
